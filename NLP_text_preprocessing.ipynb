```python
import pandas as pd

data = pd.read_csv(
    "twitter_training.csv",
    header=None,
    names=["tweet_id", "entity", "sentiments", "reviews"]
)

data.head()
```

```python
data.drop(["tweet_id", "entity"], axis=1, inplace=True)

data.columns = ["sentiments", "reviews"]

data.head()
```

```python
data.drop_duplicates(inplace=True)

data.dropna(inplace=True)

data.isna().sum()
```

```python
data.sentiments.value_counts()
```

```python
import warnings
warnings.filterwarnings("ignore")

import string
import re

from nltk.corpus import stopwords
import nltk

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))
```

```python
def preprocessor(text):

    # Lowercase
    text = text.lower()

    # Remove punctuation
    text = "".join([
        a for a in text
        if a not in string.punctuation
    ])

    # Remove stopwords
    text = " ".join([
        a for a in text.split()
        if a not in stop_words
    ])

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    return text
```

```python
preprocessor("HELLOOO 6789 This is an AMAZING game!!!")
```

```python
data["reviews"] = data["reviews"].apply(preprocessor)

data.head(10)
```

```python
data.to_csv("twitter_preprocessed.csv", index=False)

print("Preprocessing completed successfully!")
```
